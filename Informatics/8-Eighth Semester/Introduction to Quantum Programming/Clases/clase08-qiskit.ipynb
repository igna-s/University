**Normalizando un Vector de Estado**

Vimos que el vector $|v \rangle=[1, 1]$ no es un estado cuántico válido porque su norma no es 1. Podemos normalizarlo para obtener un  `Statevector` válido. Para ello:

1. Calculamos la **norma** del vector $| v \rangle$.
2. Creamos un nuevo vector dividiendo al vector $| v \rangle$ por su norma.

$$ \text{vector normalizado} = \frac{1}{\|v\|} |v \rangle $$

Sabemos que $\|v\|^2 = \langle v, v \rangle$ . En qiskit con `v.inner(v)` obtenemos el producto interno de $v$ consigo mismo. Lo usaremos para calcular la norma.

Una vez que tenemos la norma, simplemente dividimos al  `Statevector` por la norma.


````python
# Calculamos la norma v
norma = sqrt(v.inner(v))

# Dividimos el Statevector por la norma
v_normalizado = v / norma

# Verificamos
print("Vector normalizado:")
display(v_normalizado.draw('latex'))
print(f"¿Es válido? {v_normalizado.is_valid()}")
````

### Realizar operaciones con `Operator` y `Statevector`

Las operaciones unitarias pueden definirse en Qiskit utilizando la clase `Operator`, como en el ejemplo que sigue.
Esta clase incluye un método `draw` con argumentos similares a `Statevector`.

````python
# Reiniciamos el entorno de ejecución (borra todas las variables) para comenzar esta sección con un entorno limpio
%reset -f

from qiskit.quantum_info import Operator, Statevector #importamos las clases Operator y Statevector
from numpy import sqrt  #importamos la función sqrt de la biblioteca NumPy para calcular una raíz cuadrada

H = Operator([[1 / sqrt(2), 1 / sqrt(2)], [1 / sqrt(2), -1 / sqrt(2)]])
X = Operator([[0, 1], [1, 0]])  
Y = Operator([[0, -1.0j], [1.0j, 0]])
Z = Operator([[1, 0], [0, -1]]) # rotación de π alrededor del eje Z

S = Operator([[1, 0], [0, 1.0j]]) # rotación de π/2 alrededor del eje Z
T = Operator([[1, 0], [0, (1 + 1.0j) / sqrt(2)]]) # rotación de π/4 alrededor del eje Z

display(H.draw("latex"))
display(X.draw("latex"))
display(Y.draw("latex"))
display(Z.draw("latex"))

display(S.draw("latex"))
display(T.draw("latex"))
````

**Verificación de Propiedades de Operadores**

La clase `Operator` incluye métodos para verificar sus propiedades matemáticas.
- **`.is_unitary()`**: Comprueba si el operador es unitario ($U^\dagger U = I$). Todas las compuertas cuánticas válidas deben cumplir esta condición.

Para la **hermiticidad**, no hay un método directo. La comprobamos usando la definición: un operador $\hat{A}$ es hermitiano si es igual a su propio adjunto ($\hat{A}^\dagger = \hat{A}$). En Qiskit, el adjunto se calcula con el método **`.adjoint()`**.

Vamos a probar esto con algunos de nuestros operadores.

````python
# Verifiquemos la unitariedad de la compuerta Hadamard (H)
print(f"¿Es H unitaria? {H.is_unitary()}")

# Para verificar la hermiticidad, comparamos el operador con su adjunto
# Un operador A es hermitiano si A == A.adjoint()

# Verifiquemos la hermiticidad del operador Pauli-Z (Z)
is_Z_hermitian = (Z == Z.adjoint())
print(f"¿Es Z hermitiana? {is_Z_hermitian}")

# Una compuerta puede ser unitaria y hermitiana a la vez
is_H_hermitian = (H == H.adjoint())
print(f"¿Es H hermitiana? {is_H_hermitian}")
````

Ahora, veamos un ejemplo de una compuerta que es unitaria pero **no** hermitiana, como la compuerta S.

````python
# La compuerta S es una rotación de fase, por lo que debe ser unitaria
print(f"¿Es S unitaria? {S.is_unitary()}")

# Ahora, comprobemos su hermiticidad
is_S_hermitian = (S == S.adjoint())
print(f"¿Es S hermitiana? {is_S_hermitian}")

# Mostremos las matrices para entender por qué
print("\nMatriz de S:")
display(S.draw('latex'))
print("Matriz adjunta de S (S†):")
display(S.adjoint().draw('latex'))

# Como S != S†, la compuerta S no es hermitiana.
````


### Producto tensorial

Las clases `Statevector` y `Operator` tienen un método `tensor` (y su operador equivalente `^`) para calcular el producto tensorial. Este método acepta otro `Statevector` u `Operator` como argumento, que se interpreta como el factor tensorial de la derecha.

El resultado será un `Statevector` solo si ambos factores lo son. Si al menos uno de los factores es un `Operator`, el resultado será un `Operator`.

A continuación creamos dos vectores de estado que representan $\vert 0\rangle$ y $\vert 1\rangle,$ y utilizamos el método `tensor` primero y el operador `^` después para obtener $\vert \psi\rangle = \vert 0\rangle \otimes \vert 1\rangle$ y $\vert \phi\rangle = \vert 1\rangle \otimes \vert 0\rangle$.  
Observar que estamos utilizando el método `from_label` para definir los estados $\vert 0\rangle$ y $\vert 1\rangle,$ en lugar de definirlos nosotros mismos.


````python
# Reiniciamos el entorno de ejecución (borra todas las variables) para comenzar esta sección con un entorno limpio
%reset -f

from qiskit.quantum_info import Statevector, Operator

zero = Statevector.from_label("0")
one = Statevector.from_label("1")
psi = zero.tensor(one)
phi = one ^ zero
display(psi.draw("latex"))
display(phi.draw("latex"))
````

Otras etiquetas permitidas son "+" y "-" para los estados más y menos, así como "r" y "l" (abreviatura de "derecha" e "izquierda") para los estados 

$$
\vert {+i} \rangle = \frac{1}{\sqrt{2}} \vert 0 \rangle + \frac{i}{\sqrt{2}} \vert 1 \rangle
\qquad\text{y}\qquad
\vert {-i} \rangle = \frac{1}{\sqrt{2}} \vert 0 \rangle - \frac{i}{\sqrt{2}} \vert 1 \rangle.
$$

He aquí un ejemplo del producto tensorial de $\vert {+} \rangle$ y $\vert {-i} \rangle.$



````python
plus = Statevector.from_label("+")
minus_i = Statevector.from_label("l")
phi = plus.tensor(minus_i)
display(phi.draw("latex"))
````

Ejemplo de producto tensorial entre operadores. La clase `Operator` de Qiskit también posee un método from_label() para crear operadores comunes.



````python
H = Operator.from_label("H")
Id = Operator.from_label("I")
X = Operator.from_label("X")
display(H.tensor(Id).draw("latex"))
display(H.tensor(Id).tensor(X).draw("latex"))
# también podemos usar el operador ^ 
display((H ^Id ^ X).draw("latex"))
````

Los estados compuestos pueden evolucionar utilizando operaciones compuestas.  
Por ejemplo, el siguiente código calcula el estado $(H\otimes I)\vert\phi\rangle$ para $\vert\phi\rangle = \vert + \rangle \otimes \vert {-i}\rangle$ (que ya se definió anteriormente).

````python
display(phi.evolve(H ^ Id).draw("latex"))
````

El resultado es el estado de Bell $\vert\phi^{+}\rangle.$

### Medidas parciales

Ya vimos el método `measure` para simular una medición de un vector de estado cuántico.
Este método devuelve dos elementos: el resultado de la medición simulada y el nuevo `Statevector` dada esta medición.

Por defecto, `measure` mide todos los qubits del vector de estados.
Alternativamente, podemos proporcionar una lista de enteros como argumento, lo que hace que sólo se midan esos índices de qubits.
Para demostrarlo, el código siguiente crea el estado

$$
\vert w\rangle = \frac{\vert 001\rangle + \vert 010\rangle + \vert 100\rangle}{\sqrt{3}}
$$

y mide el qubit número 0, que es el qubit más a la derecha.




````python
# Reiniciamos el entorno de ejecución (borra todas las variables) para comenzar esta sección con un entorno limpio
%reset -f

from qiskit.quantum_info import Statevector
from numpy import sqrt

w = Statevector([0, 1, 1, 0, 1, 0, 0, 0] / sqrt(3)) # El estado W es |001⟩ + |010⟩ + |100⟩.
display(w.draw("latex"))

result, state = w.measure([0])
print(f"Measured: {result}\nState after measurement:")
display(state.draw("latex"))

result, state = w.measure([0, 1])
print(f"Measured: {result}\nState after measurement:")
display(state.draw("latex"))

result, state = w.measure([0, 2])
print(f"Measured: {result}\nState after measurement:")
display(state.draw("latex"))


#Observar que `result` muestra los bits medidos en el orden de qiskit (los de mayor índice a la izquierda)
# si medimos los índices [0,2] y obtenemos '10' significa que q_2 es 1 y q_0 es 0
````

### Construyendo Circuitos de 1 Qubit con Qiskit

En la sección anterior, trabajamos con `Statevector` y `Operator` para manipular estados y compuertas de forma algebraica. Ahora, vamos a dar el siguiente paso y aprender a construir **circuitos cuánticos**, que es la forma estándar de describir algoritmos en Qiskit.

Utilizaremos la clase `QuantumCircuit` para ensamblar secuencias de compuertas y luego analizaremos el resultado.

Utilizaremos el método `draw` de la clase `QuantumCircuit` con el renderizador `mpl` (abreviatura de `Matplotlib`, una biblioteca de visualización Python) para visualizar el circuito.

````python
# Reiniciamos el entorno de ejecución (borra todas las variables) para comenzar esta sección con un entorno limpio
%reset -f

# Importaciones fundamentales para esta sección
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector, Operator
from qiskit.visualization import plot_histogram
from numpy import sqrt
````

Como vimos en la teoría, la compuerta `NOT`, también conocida como compuerta `X` realiza un "bit-flip".   
Vamos a construir un circuito simple que aplica una compuerta `X` al único qubit del circuito.

````python
# 1. Creamos un circuito con un solo qubit
qc_x = QuantumCircuit(1)

# 2. Aplicamos la compuerta X al qubit en el índice 0
qc_x.x(0)

# 3. Dibujamos el circuito
display(qc_x.draw("mpl"))
````

**Obteniendo el Operador de un Circuito**

Una vez que hemos construido un `QuantumCircuit`, a menudo queremos analizar la matriz unitaria total que representa. Para ello, Qiskit nos ofrece dos formas equivalentes de crear un objeto `Operator` a partir de un `QuantumCircuit`.

1.  **Usando el método de clase `.from_circuit()`:** Esta es la forma más explícita y recomendada para empezar, ya que deja clara la intención del código.
2.  **Usando el constructor `Operator()`:** Este es un atajo más "Pythónico" y conciso.

Las siguientes dos líneas de código producen exactamente el mismo resultado:

```python
# Opción 1 (Explícita y recomendada)
op_from_method = Operator.from_circuit(mi_circuito)

# Opción 2 (Concisa)
op_from_constructor = Operator(mi_circuito)

````python
# Verifiquemos que la matriz del circuito es la matriz X
op_x = Operator.from_circuit(qc_x)

print("Matriz del circuito:")
display(op_x.draw("latex"))
````

**La Compuerta Hadamard (H)**

La compuerta Hadamard (`qc.h()`) es nuestra herramienta fundamental para crear superposiciones. Vamos a verificar su acción sobre los estados base.

````python
# Creamos un circuito solo con la compuerta H
qc_h = QuantumCircuit(1)
qc_h.h(0)

# Estado inicial |0>
ket0 = Statevector.from_label('0')

# Estado final después de aplicar el circuito
estado_plus = ket0.evolve(qc_h)

print("Circuito con H:")
display(qc_h.draw("mpl"))
print("\nResultado de aplicar H a |0>:")
display(estado_plus.draw("latex"))

# Ejercicio rápido: ¿Qué obtendrías si el estado inicial fuera |1>?
# ket1 = Statevector.from_label('1')
# estado_minus = ket1.evolve(qc_h)
# display(estado_minus.draw("latex"))
````

**Añadiendo Compuertas Arbitrarias: `qc.unitary()`**

Ya conocemos los "atajos" para las compuertas comunes (`qc.h()`, `qc.x()`, etc.). Pero, ¿cómo añadimos una compuerta definida por una matriz unitaria cualquiera que no está en la biblioteca estándar?

La forma más directa es con el método `.unitary()` de un objeto `QuantumCircuit`.

**Sintaxis:** `qc.unitary(operador, qubits, label='mi_etiqueta')`

-   **`operador`**: La matriz unitaria, que puede ser una lista de listas de Python o un array de NumPy.
-   **`qubits`**: Una **lista** con los índices de los qubits a los que se aplica la compuenta.
-   **`label`**: (Opcional) Un nombre para la compuerta que se mostrará en el diagrama del circuito.

Vamos a verificar este método recreando una compuerta que ya conocemos: la Hadamard (H).

````python
# 1. Creamos el circuito
qc = QuantumCircuit(1)

# 2. Definimos la matriz de Hadamard
H_matrix = [
    [1/sqrt(2),  1/sqrt(2)],
    [1/sqrt(2), -1/sqrt(2)]
]

# 3. La agregamos al qubit 0 con una etiqueta personalizada
qc.unitary(H_matrix, [0], label='H_manual')

# 4. Dibujamos el circuito
print("Circuito con nuestra compuerta H manual:")
display(qc.draw('mpl'))

# 5. Verificamos que el operador del circuito corresponde a la matriz H
op_manual = Operator(qc)
print("\nMatriz del operador resultante:")
display(op_manual.draw('latex'))
````

**La Puerta de Fase Parametrizada (P)**

La compuerta P (`qc.p()`) nos permite aplicar una fase $e^{i\phi}$ al estado $|1\rangle$. A diferencia de las anteriores, esta compuerta necesita un **parámetro**: el ángulo $\phi$ en radianes.

Sus casos particulares más importantes son `S`, `T` y `Z`.

````python
from numpy import pi 

# --- Compuerta S ---
# Construimos S usando P(π/2)
qc_s_from_p = QuantumCircuit(1)
qc_s_from_p.p(pi/2, 0)
op_s_from_p = Operator(qc_s_from_p)

# --- Compuerta T ---
# Construimos T usando P(π/4)
qc_t_from_p = QuantumCircuit(1)
qc_t_from_p.p(pi/4, 0)
op_t_from_p = Operator(qc_t_from_p)

# --- Compuerta Z ---
# Construimos Z usando P(π)
qc_z_from_p = QuantumCircuit(1)
qc_z_from_p.p(pi, 0)
op_z_from_p = Operator(qc_z_from_p)

# Mostramos las matrices para verificar
print("Matriz de S, construida como P(π/2):")
display(op_s_from_p.draw('latex'))
# Para comparar, la compuerta nativa S de Qiskit
# print("Matriz de la compuerta nativa S:")
# display(Operator.from_label('S').draw('latex'))

print("\nMatriz de T, construida como P(π/4):")
display(op_t_from_p.draw('latex'))

print("\nMatriz de Z, construida como P(π):")
display(op_z_from_p.draw('latex'))
````

**Verificando Identidades de Compuertas en Qiskit**

La teoría nos dice que existen relaciones interesantes entre las compuertas, como $Z = S^2$ o $X=HZH$. Podemos verificar estas identidades construyendo los circuitos correspondientes y comparando sus operadores.

````python
# Circuito para S^2 (S seguida de S)
qc_s_cuadrado = QuantumCircuit(1)
qc_s_cuadrado.s(0)
qc_s_cuadrado.s(0)

# Obtenemos el operador del circuito
op_s_cuadrado = Operator.from_circuit(qc_s_cuadrado)

print("Circuito para S * S:")
display(qc_s_cuadrado.draw('mpl'))

print("\nOperador resultante de S * S:")
display(op_s_cuadrado.draw('latex'))

print("¿Es igual a la compuerta Z?")
print(op_s_cuadrado == Operator.from_label('Z'))
````

**Visualización en la Esfera de Bloch: Siguiendo la Evolución del Estado**

Para ganar una intuición geométrica de la identidad $S^2=Z$, vamos a "seguir" un vector de estado a medida que pasa a través de nuestro circuito $S^2$.

Partiremos del estado inicial $|+\rangle$ (que no está en el eje de rotación Z) y visualizaremos en la Esfera de Bloch cada paso de la evolución:
1. El estado inicial: $|+\rangle$.
2. El estado intermedio: $S|+\rangle$.
3. El estado final: $S(S|+\rangle) = S^2|+\rangle$.

Finalmente, podremos observar que el punto final de esta "trayectoria" es el mismo que si hubiésemos aplicado la compuerta `Z` directamente.

````python
from qiskit.visualization import plot_bloch_multivector

# Estado inicial
estado = Statevector.from_label('+')
print("Estado Inicial:")
display(plot_bloch_multivector(estado)) 

# Bucle para evolucionar y mostrar paso a paso 
for idx, instruction in enumerate(qc_s_cuadrado.data):
    # Extraemos la compuerta de la instrucción
    gate_to_apply = instruction.operation
    # Evolucionamos el estado con la operación actual
    estado = estado.evolve(gate_to_apply)

    print(f"\nDespués del Paso {idx} (aplicando '{gate_to_apply.name}'):")
    display(plot_bloch_multivector(estado))
````

````python
# Circuito para HZH
qc_hzh = QuantumCircuit(1)
qc_hzh.h(0)
qc_hzh.z(0)
qc_hzh.h(0)

# Obtenemos el operador y lo comparamos con X
op_hzh = Operator(qc_hzh)

print("Circuito para HZH:")
display(qc_hzh.draw('mpl'))
print("\nOperador resultante de HZH:")
display(op_hzh.draw('latex'))
print("¿Es igual a la compuerta X?")
print(op_hzh == Operator.from_label('X'))
````

Vamos a "seguir" el vector de estado $|0\rangle$ a medida que pasa a través de nuestro circuito $HZH$.

Finalmente, podremos observar que el punto final de esta "trayectoria" es el mismo que si hubiésemos aplicado la compuerta `X` directamente.

````python
# Estado inicial
estado = Statevector.from_label('0')
print("Estado Inicial:")
display(plot_bloch_multivector(estado)) 

# Bucle para evolucionar y mostrar paso a paso 
for idx, instruction in enumerate(qc_hzh.data):
    # Extraemos la compuerta de la instrucción
    gate_to_apply = instruction.operation
    # Evolucionamos el estado con la operación actual
    estado = estado.evolve(gate_to_apply)

    print(f"\nDespués del Paso {idx} (aplicando '{gate_to_apply.name}'):")
    display(plot_bloch_multivector(estado))
````

**Alizando Resultados**  
Construiremos un circuito, obtendremos su operador, haremos evolucionar un estado y simularemos mediciones.

````python
# 1. Construir el circuito
circuit = QuantumCircuit(1)
circuit.h(0)
circuit.s(0)
circuit.h(0)
print("Circuito Completo:")
display(circuit.draw("mpl"))

# 2. Obtener su operador unitario total
U_total = Operator.from_circuit(circuit)
print("\nMatriz unitaria del circuito:")
display(U_total.draw("latex"))

# 3. Hacer evolucionar el estado |0>
ket0 = Statevector([1,0])
estado_final = ket0.evolve(circuit)
print("\nEstado final al aplicar el circuito a |0>:")
display(estado_final.draw("latex"))

# Bucle para evolucionar y mostrar paso a paso
for idx, instruction in enumerate(circuit.data):
    # Extraemos la compuerta de la instrucción
    gate_to_apply = instruction.operation
    # Evolucionamos el estado con la operación actual
    ket0 = ket0.evolve(gate_to_apply)

    print(f"\nDespués del Paso {idx} (aplicando '{gate_to_apply.name}'):")
    display(plot_bloch_multivector(ket0))
````

### Construyendo Circuitos de Múltiples Qubits con Qiskit

En esta sección, aprenderemos a:
1. Crear un `QuantumCircuit` con múltiples "cables".
2. Aplicar compuertas a qubits específicos dentro del circuito.
3. Analizar el estado global del sistema resultante.

````python
# Reiniciamos el entorno para asegurar que empezamos desde cero
%reset -f

# Importaciones fundamentales para esta sección
import numpy as np
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector, Operator
from qiskit.visualization import plot_histogram, plot_bloch_multivector
````

**Creando un Circuito Multi-Qubit**

Para crear un circuito con múltiples qubits, simplemente pasamos el número de qubits deseado al constructor de `QuantumCircuit`.

````python
# Creamos un circuito con 2 qubits
qc = QuantumCircuit(2)

# Aplicamos una compuerta H al qubit 1 (el cable de abajo)
qc.h(1)

# El qubit 0 (el cable de arriba) queda sin tocar, lo que equivale a aplicarle la Identidad (I)

# Dibujamos el circuito
print("Circuito para H ⊗ I:")
display(qc.draw('mpl'))

# Verifiquemos su efecto sobre el estado inicial |00>
ket00 = Statevector.from_label('00')
estado_final = ket00.evolve(qc)

print("\nEstado inicial |00>:")
display(ket00.draw('latex'))

print("\nEstado final después de aplicar H ⊗ I:")
display(estado_final.draw('latex'))
print("\nComo vimos en la teoría, el resultado es (1/√2)(|00> + |10>)")
````

**Ejemplo Práctico: Creando una Superposición Uniforme**

Uno de los patrones más comunes en los algoritmos cuánticos es aplicar una compuerta Hadamard a cada qubit para crear una superposición uniforme de todos los estados base. Vamos a construir el circuito para 3 qubits y a verificar el estado final.

````python
# 1. Creamos un circuito con 3 qubits
qc = QuantumCircuit(3)

# 2. Aplicamos una compuerta Hadamard a cada qubit
# Podemos hacerlo individualmente...
# qc.h(0)
# qc.h(1)
# qc.h(2)
# ...o de forma más compacta con una lista de qubits:
qc.h([0, 1, 2])

# 3. Dibujamos el circuito
print("Circuito para crear la superposición uniforme de 3 qubits (H^⊗3):")
display(qc.draw('mpl'))

# 4. Calculamos el estado final a partir de |000>
estado_inicial = Statevector.from_label('000')
estado_final = estado_inicial.evolve(qc)

# 5. Mostramos el estado final
print("\nEstado final resultante:")
display(estado_final.draw('latex'))

print("\nComo se puede ver, todas las amplitudes son iguales (1/√8), confirmando la superposición uniforme.")
````

**Combinando Operadores en Qiskit**

La clase `Operator` nos permite combinar compuertas de dos formas principales, que se corresponden con las reglas de traducción de circuitos:

1.  **Composición (Producto Matricial):** Para compuertas en secuencia en el mismo cable. Se utiliza el operador `@`.
    ```python
    U_total_cable = U_final @ U_inicial
    ```
2.  **Producto Tensorial:** Para compuertas simultáneas en cables diferentes. Se utiliza el operador `^`.
    ```python
    U_total_columna = U_q1 ^ U_q0
    ```
