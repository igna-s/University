## Sección 1: El "Copiador" Clásico vs. La Realidad Cuántica

En la computación clásica, la operación de copiar información (Ctrl+C) es fundamental y la damos por sentada. En esta sección, vamos a investigar si podemos construir una "máquina de clonación" similar en el mundo cuántico.

Nuestra principal "sospechosa" para ser una compuerta copiadora es la **CNOT**. A primera vista, su comportamiento parece el de una copiadora: si el qubit objetivo está en el estado $|0\rangle$, parece copiar el estado del qubit de control.

Vamos a verificar este comportamiento para los estados de la base computacional.

```python
# Reiniciamos el entorno para empezar desde cero
%reset -f

# Importaciones necesarias
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
```

```python
# 1. Creamos un circuito con la compuerta CNOT (control en q1, objetivo en q0)
qc = QuantumCircuit(2)
qc.cx(1, 0)

# --- Caso 1: Intentando copiar el estado |0> ---
estado_inicial_0 = Statevector.from_label('00')
estado_final_0 = estado_inicial_0.evolve(qc)

print("--- Caso 1: Copiando un |0> ---")
print("Circuito aplicado:")
display(qc.draw('mpl'))
print(f"Estado Inicial: |ψ⟩ = |0⟩|0⟩")
display(estado_inicial_0.draw('latex'))
print("Estado Final:")
display(estado_final_0.draw('latex'))
print("Resultado: El estado no cambia. El |0> del control parece haberse 'copiado'.")

print("\n" + "-"*50 + "\n")

# --- Caso 2: Intentando copiar el estado |1> ---
estado_inicial_1 = Statevector.from_label('10') # |q1>|q0> = |1>|0>
estado_final_1 = estado_inicial_1.evolve(qc)

print("--- Caso 2: Copiando un |1> ---")
print(f"Estado Inicial: |ψ⟩ = |1⟩|0⟩")
display(estado_inicial_1.draw('latex'))
print("Estado Final:")
display(estado_final_1.draw('latex'))
print("Resultado: El estado del control |1> se ha 'copiado' al objetivo, resultando en |11>.")
```

**Conclusión**

Hasta ahora, parece que la CNOT es una copiadora perfecta. Hemos verificado que para la base computacional, `CNOT(|x⟩|0⟩) = |x⟩|x⟩`.

Sin embargo, el **Teorema de No-Clonación** se aplica a estados *arbitrarios y desconocidos*, lo que incluye las superposiciones. 

En la siguiente sección, pondremos a nuestra compuerta "copiadora" a prueba de verdad, intentando clonar un estado en superposición.

**Intentando Clonar una Superposición**

Ahora vamos a realizar el experimento crucial. En lugar de darle a nuestra CNOT un estado base conocido como $|0\rangle$ o $|1\rangle$, le daremos un estado en superposición: el estado $|+\rangle$.

El experimento consiste en preparar el sistema en el estado inicial $|+\rangle|0\rangle$ y aplicarle una CNOT con control en el primer qubit (el que está en $|+\rangle$).

Tenemos dos posibles resultados:
1.  **Resultado de Clonación Ideal:** Si la CNOT fuera un clonador universal, el estado final debería ser dos copias independientes de $|+\rangle$, es decir, el estado separable $|+\rangle|+\rangle$.
2.  **Resultado de la Evolución Cuántica Real:** El estado que realmente obtendremos al simular la operación en Qiskit.

Vamos a calcular ambos y compararlos.

```python
# RESOLUCIÓN DEL EJERCICIO 3

%reset -f


from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
import numpy as np # Importamos numpy para usar su función 'isclose'

# --- Generación del Estado Ideal (Resultado de la Clonación) ---
estado_plus = Statevector.from_label('+')
estado_ideal = estado_plus.tensor(estado_plus)

print("--- 1. Resultado de una CLONACIÓN IDEAL ---")
print("Estado |ψ_ideal> = |+> ⊗ |+>:")
display(estado_ideal.draw('latex'))


# --- Generación del Estado Real (Resultado de la Evolución Cuántica) ---
estado_cero = Statevector.from_label('0')
estado_inicial_real = estado_plus.tensor(estado_cero)

qc = QuantumCircuit(2)
qc.cx(1, 0) # Control en q1 (el |+>), objetivo en q0 (el |0>)

estado_real = estado_inicial_real.evolve(qc)

print("\n--- 2. Resultado de la EVOLUCIÓN CUÁNTICA REAL ---")
print("Estado |ψ_real> = CNOT(|+> ⊗ |0>):")
display(estado_real.draw('latex'))


# --- Comparación y Análisis ---
print("\n--- 3. COMPARACIÓN Y ANÁLISIS ---")

# Comparación directa de los vectores de estado
son_iguales = (estado_ideal == estado_real)
print(f"¿Son iguales el estado ideal y el estado real? -> {son_iguales}")

print("-" * 50)

# Verificación de la separabilidad aplicando el test teórico
print("Verificando la separabilidad con el test c₀₀*c₁₁ == c₀₁*c₁₀:")

# NOTA: Usamos np.isclose() en lugar de '==' para comparar números de punto flotante.
# Esto evita problemas con pequeños errores de redondeo que son inherentes a cómo
# las computadoras manejan decimales. np.isclose(a, b) comprueba si 'a' y 'b' son
# aproximadamente iguales (dentro de una tolerancia muy pequeña).

# Test para el estado ideal
c00_i, c01_i, c10_i, c11_i = estado_ideal.data
es_separable_ideal = np.isclose(c00_i * c11_i, c01_i * c10_i)
print(f"¿Es separable el estado ideal (|ψ_ideal>)? -> {es_separable_ideal}")

# Test para el estado real
c00_r, c01_r, c10_r, c11_r = estado_real.data
es_separable_real = np.isclose(c00_r * c11_r, c01_r * c10_r)
print(f"¿Es separable el estado real (|ψ_real>)? -> {es_separable_real}")
```

**Conclusión**

El resultado es claro:
- El estado idealmente clonado, $|+\rangle|+\rangle$, es **separable**.
- El estado real que produce la CNOT, $\frac{1}{\sqrt{2}}(|00\rangle+|11\rangle)$, es **entrelazado**.

En lugar de crear dos copias independientes (clonación), crea correlaciones cuánticas (entrelazamiento).

> **La CNOT no clona, entrelaza.**

## Sección 2 - Medición en Distintas Bases

### El Flujo de Trabajo de Simulación en Qiskit

Hasta ahora, hemos trabajado principalmente con la clase `Statevector`, que es excelente para analizar la evolución matemática de los estados. Sin embargo, para simular circuitos completos que incluyen mediciones (como lo haríamos en un ordenador cuántico real), necesitamos un nuevo paradigma.

El flujo de trabajo estándar en Qiskit consta de 3 pasos:

1.  **Construir:** Se crea un `QuantumCircuit` que describe el algoritmo, incluyendo las compuertas y las mediciones finales.
2.  **Ejecutar:** Se elige un `backend` (un simulador o un hardware cuántico real) y se envía el circuito para su ejecución. Esto crea un `Job`.
3.  **Analizar:** Se recogen los resultados del `Job`, que típicamente son un diccionario de conteos (`counts`).

Para la simulación, usaremos el `AerSimulator` de Qiskit.

```python
# Si no está instalado, es necesario instalar qiskit_aer
%pip install qiskit_aer
```

```python
# Reiniciamos el entorno para empezar desde cero
%reset -f

# Si no están instalados, es necesario instalar qiskit y qiskit_aer
# %pip install qiskit
# %pip install qiskit_aer

# Importaciones necesarias para la simulación
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator # El simulador
from qiskit.visualization import plot_histogram

# 1. CONSTRUIR: Un circuito simple para demostrar el flujo
# Preparamos un estado |+> y lo medimos
qc = QuantumCircuit(1, 1) # 1 qubit, 1 bit clásico
qc.h(0)
qc.measure(0, 0) # Mide el qubit 0 y guarda en el bit clásico 0
print("Circuito de demostración:")
display(qc.draw('mpl'))

# 2. EJECUTAR: Configuramos y ejecutamos el simulador
simulator = AerSimulator()
# El paso de 'transpile' optimiza el circuito para el backend. Es una buena práctica.
compiled_circuit = transpile(qc, simulator)
job = simulator.run(compiled_circuit, shots=1000) # shots = número de repeticiones

# 3. ANALIZAR: Obtenemos los resultados
result = job.result()
counts = result.get_counts() 
print("\nResultados de la simulación (counts):")
print(counts)
print("\nHistograma de resultados:")
display(plot_histogram(counts))

```

### Medición en Diferentes Bases

En teoría, hemos visto que podemos medir un qubit con respecto a diferentes observables (como Pauli-X o Pauli-Y), lo que equivale a medir en diferentes bases. Sin embargo, en la práctica, los ordenadores cuánticos (y nuestros simuladores) están diseñados para realizar mediciones únicamente en **una base**: la **base computacional (Z)**.

**El "Truco" del Cambio de Base**

Para medir en una base diferente, usamos una estrategia ingeniosa: aplicamos una transformación unitaria *justo antes* de la medición en Z. Esta transformación rota el estado de tal manera que la medición en Z nos da la información que buscábamos en la otra base.

**Nuestras Recetas:**
-   **Para Medir en Base X (Hadamard):**
    1.  Aplicar una compuerta `H`.
    2.  Medir en la base Z.
    *Un resultado de `'0'` corresponde a haber medido $|+\rangle$, y un `'1'` a $|-\rangle$.*

-   **Para Medir en Base Y (Circular):**
    1.  Aplicar una compuerta `Sdg` (S†).
    2.  Aplicar una compuerta `H`.
    3.  Medir en la base Z.
    *Un resultado de `'0'` corresponde a haber medido $|i\rangle$, y un `'1'` a $|-i\rangle$.*

En los siguientes ejercicios resueltos, verificaremos estas recetas en la práctica.

```python
# Reiniciamos el entorno de ejecución para empezar desde cero
%reset -f

#------------------------------------------------------------------
# Ejercicio 1 resuelto: Verificando la Medición en Base X
#------------------------------------------------------------------
# Instrucciones:
# Vamos a preparar el estado |+> y luego medirlo en la base Z y en la base X
# para observar la diferencia.

# --- Importaciones y Setup ---
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram

simulator = AerSimulator()
shots = 1024

# --- Experimento 1: Preparar |+> y medir en Base Z ---
qc_Z = QuantumCircuit(1, 1)

# Preparación del estado
qc_Z.barrier(label="Preparo |+>") # Separador visual
qc_Z.h(0)

# Medición
qc_Z.barrier(label="Mido en Z") # Separador visual
qc_Z.measure(0, 0)

print("--- Medición del estado |+> en la Base Z ---")
display(qc_Z.draw('mpl'))
job_z = simulator.run(transpile(qc_Z, simulator), shots=shots)
counts_z = job_z.result().get_counts()
print("Resultados (esperado: 50% '0', 50% '1'):")
display(plot_histogram(counts_z))


# --- Experimento 2: Preparar |+> y medir en Base X ---
qc_X = QuantumCircuit(1, 1)
# Preparación del estado
qc_X.barrier(label="Preparo |+>")
qc_X.h(0)

# Cambio de base para la medición
qc_X.barrier(label="Mido en X")
qc_X.h(0)

# Medición
qc_X.measure(0, 0)

print("\n--- Medición del estado |+> en la Base X ---")
display(qc_X.draw('mpl'))
job_x = simulator.run(transpile(qc_X, simulator), shots=shots)
counts_x = job_x.result().get_counts()
print("Resultados (esperado: 100% '0'):")
display(plot_histogram(counts_x))
```

```python
# Reiniciamos el entorno de ejecución para empezar desde cero
%reset -f

#------------------------------------------------------------------
# Ejercicio 2 resuelto: Verificando la Medición en Base Y
#------------------------------------------------------------------
# Instrucciones:
# Ahora, prepara el estado |i> y mídelo en la base Z y en la base Y.

# --- Importaciones y Setup ---
from qiskit import QuantumCircuit, transpile
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram

simulator = AerSimulator()
shots = 1024

# --- Experimento 1: Preparar |i> y medir en Base Z ---
qc_z_meas_i = QuantumCircuit(1, 1)
# Preparación del estado |i> (H seguida de S)
qc_z_meas_i.barrier(label="Preparo |i>")
qc_z_meas_i.h(0)
qc_z_meas_i.s(0)
qc_z_meas_i.barrier(label="Mido en Z")
# Medición
qc_z_meas_i.measure(0, 0)

print("--- Medición del estado |i> en la Base Z ---")
display(qc_z_meas_i.draw('mpl'))
job_z_i = simulator.run(transpile(qc_z_meas_i, simulator), shots=shots)
counts_z_i = job_z_i.result().get_counts()
print("Resultados (esperado: 50% '0', 50% '1'):")
display(plot_histogram(counts_z_i))


# --- Experimento 2: Preparar |i> y medir en Base Y ---
qc_y_meas_i = QuantumCircuit(1, 1)
# Preparación del estado |i>
qc_y_meas_i.barrier(label="Preparo |i>")
qc_y_meas_i.h(0)
qc_y_meas_i.s(0)
# Separador visual
qc_y_meas_i.barrier(label="Mido en Y")
# Cambio de base para la medición en Y (Sdg y luego H)
qc_y_meas_i.sdg(0)
qc_y_meas_i.h(0)
# Medición
qc_y_meas_i.measure(0, 0)

print("\n--- Medición del estado |i> en la Base Y ---")
display(qc_y_meas_i.draw('mpl'))
job_y_i = simulator.run(transpile(qc_y_meas_i, simulator), shots=shots)
counts_y_i = job_y_i.result().get_counts()
print("Resultados (esperado: 100% '0'):")
display(plot_histogram(counts_y_i))
```

## Sección 3: Simulación del Protocolo BB84


Ahora que entendemos la teoría del protocolo BB84, vamos a simularlo usando Qiskit. Veremos cómo Alice y Bob pueden establecer una clave secreta y, lo más importante, cómo pueden detectar la presencia de una espía (Eva) gracias a las leyes de la física cuántica.

**El Plan:**
1.  **Ejercicio 1 (Canal Seguro):** Simularemos el protocolo sin la presencia de Eva para verificar que Alice y Bob terminan con una clave idéntica.
2.  **Ejercicio 2 (Canal Inseguro):** Repetiremos la simulación, pero esta vez con Eva interceptando, midiendo y reenviando cada qubit. Veremos cómo su espionaje introduce errores detectables.

```python
# Reiniciamos el entorno para estar seguros de que empezamos desde cero
%reset -f

# Importaciones necesarias
import numpy as np
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram

# --- FUNCIONES AUXILIARES PARA EL PROTOCOLO ---

def crear_circuito_alice(bit, base):
    """Prepara un qubit según el bit y la base de Alice."""
    qc = QuantumCircuit(1, 1)
    if bit == 1:
        qc.x(0)  # Aplica X si el bit es 1
    if base == 'X':
        qc.h(0)  # Aplica H si la base es X
    return qc

def medir_circuito_bob(qc, base):
    """Añade las compuertas de medición de Bob según su base."""
    if base == 'X':
        qc.h(0)
    qc.measure(0, 0)
    return qc
```

**Ejercicio 1: Simulación de un Canal Seguro (Sin Eva)**

En este primer escenario, Alice enviará una secuencia de qubits a Bob sin que nadie intercepte el canal. Verificaremos que, después del sondeo de bases, la clave que les queda es idéntica.

```python
# --- 1. Preparación (Alice y Bob) ---
np.random.seed(41) # Fijamos la semilla para que los resultados sean reproducibles
num_bits = 30 # Longitud de la clave inicial

# Alice genera sus bits y bases aleatorias
alice_bits = np.random.randint(2, size=num_bits)
alice_bases = np.random.choice(['Z', 'X'], size=num_bits)

# Bob genera sus bases aleatorias
bob_bases = np.random.choice(['Z', 'X'], size=num_bits)

print(f"Bits de Alice:    {alice_bits}")
print(f"Bases de Alice:   {''.join(alice_bases)}")
print(f"Bases de Bob:     {''.join(bob_bases)}")
print("-" * 30)

# --- 2. Simulación del Intercambio Cuántico ---
simulator = AerSimulator()
bob_results = []

for i in range(num_bits):
    # Alice prepara su qubit
    circuito_alice = crear_circuito_alice(alice_bits[i], alice_bases[i])
    
    # Bob añade sus compuertas de medición
    circuito_final = medir_circuito_bob(circuito_alice, bob_bases[i])
    
    # Ejecutamos la simulación (1 solo shot, porque es 1 bit)
    job = simulator.run(circuito_final, shots=1, memory=True)
    resultado = job.result().get_memory()[0]
    bob_results.append(int(resultado))

print(f"Resultados de Bob:{np.array(bob_results)}")
print("-" * 30)

# --- 3. Fase de Sondeo (Canal Público) ---
clave_alice = []
clave_bob = []

for i in range(num_bits):
    if alice_bases[i] == bob_bases[i]:
        # Si las bases coinciden, se quedan con el bit
        clave_alice.append(int(alice_bits[i])) # Convertimos el bit de numpy a un entero de Python        
        clave_bob.append(bob_results[i])

print("--- SONDEO DE BASES (Comparación Pública) ---")
print(f"Clave de Alice (filtrada): {clave_alice}")
print(f"Clave de Bob (filtrada):   {clave_bob}")

# --- 4. Verificación ---
son_identicas = (clave_alice == clave_bob)
print(f"\n¿Las claves finales son idénticas? -> {son_identicas}")
if son_identicas:
    print("¡Éxito! Alice y Bob comparten una clave secreta.")
```

**Ejercicio 2: Simulación de un Canal Inseguro (Con Eva)**

Ahora, repetiremos el proceso, pero esta vez la espía, Eva, estará en medio del canal. Por cada qubit que envía Alice, Eva lo interceptará, lo medirá en una base aleatoria y reenviará el estado colapsado a Bob. Veremos si Alice y Bob pueden detectar su presencia.

```python
# --- 1. Preparación (Alice, Bob y Eva) ---
np.random.seed(41) # Usamos la misma semilla para comparar
# (Alice y Bob usan los mismos bits y bases que antes)

# Eva genera sus bases aleatorias para espiar
eva_bases = np.random.choice(['Z', 'X'], size=num_bits)
print("--- INTERCAMBIO CON ESPIONAJE DE EVA ---")
print(f"Bases de Alice:   {''.join(alice_bases)}")
print(f"Bases de Eva:     {''.join(eva_bases)}")
print(f"Bases de Bob:     {''.join(bob_bases)}")
print("-" * 30)

# --- 2. Simulación del Intercambio con Intercepción ---
bob_results_con_eva = []

for i in range(num_bits):
    # Alice prepara y envía su qubit
    circuito_alice = crear_circuito_alice(alice_bits[i], alice_bases[i])
    
    # --- Intercepción de Eva ---
    # Eva mide el qubit de Alice en su base aleatoria
    circuito_eva = medir_circuito_bob(circuito_alice, eva_bases[i])
    job_eva = simulator.run(circuito_eva, shots=1, memory=True)
    resultado_eva = job_eva.result().get_memory()[0]
    
    # Eva reenvía un nuevo qubit basado en lo que midió
    circuito_reenviado = crear_circuito_alice(int(resultado_eva), eva_bases[i])
    # --- Fin de la intercepción ---
    
    # Bob mide el qubit que le envió Eva
    circuito_final_bob = medir_circuito_bob(circuito_reenviado, bob_bases[i])
    job_bob = simulator.run(circuito_final_bob, shots=1, memory=True)
    resultado_bob = job_bob.result().get_memory()[0]
    bob_results_con_eva.append(int(resultado_bob))

# --- 3. Fase de Sondeo ---
clave_alice_con_eva = []
clave_bob_con_eva = []

for i in range(num_bits):
    if alice_bases[i] == bob_bases[i]:
        clave_alice_con_eva.append(int(alice_bits[i]))
        clave_bob_con_eva.append(bob_results_con_eva[i])

print("--- SONDEO DE BASES (Después del Espionaje) ---")
print(f"Clave de Alice (filtrada): {clave_alice_con_eva}")
print(f"Clave de Bob (filtrada):   {clave_bob_con_eva}")

# --- 4. Verificación y Detección de Errores ---
son_identicas_con_eva = (clave_alice_con_eva == clave_bob_con_eva)
print(f"\n¿Las claves finales son idénticas? -> {son_identicas_con_eva}")

if not son_identicas_con_eva:
    # Calculamos la tasa de error (QBER)
    errores = np.sum(np.array(clave_alice_con_eva) != np.array(clave_bob_con_eva))
    tasa_error = (errores / len(clave_alice_con_eva)) * 100
    print(f"\n¡ALERTA! Se han detectado {errores} errores en la clave.")
    print(f"Tasa de Error de Bits Cuánticos (QBER): {tasa_error:.2f}%")
    print("La presencia de Eva ha sido detectada. El protocolo se aborta.")
```
